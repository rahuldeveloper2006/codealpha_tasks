## 5. Exploratory Data Analysis (EDA) :-

### 5.1 Overall Unemployment statistics

```python
avg_unemployment_india = df["Estimated Unemployment Rate (%)"].mean().round(2)
print(f"The overall average unemployment rate in india is : {avg_unemployment_india} %")
```

**Insight:** The overall average unemployment rate in India is **11.79%**. This indicates that, on average, around 11.79% of the workforce in the dataset was unemployed during the observed period.

### 5.2 Region with the Highest Unemployment Rate

```python
highest_unemployment = df.loc[df["Estimated Unemployment Rate (%)"].idxmax(),['Region','Estimated Unemployment Rate (%)']]
highest_unemployment
```

**Insight:** Puducherry recorded the highest unemployment rate in the dataset at **76.74%**. This indicates that Puducherry experienced the most severe unemployment during the observed period, possibly due to significant economic disruptions.

### 5.3 Region with the lowest Unemployment Rate 

```python
lowest_unemployment = df.loc[df["Estimated Unemployment Rate (%)"].idxmin(),["Region","Estimated Unemployment Rate (%)"]]
lowest_unemployment
```

**Insight:** Assam recorded the lowest unemployment rate in the dataset at **0.00%**, indicating that it experienced the lowest observed unemployment during the recorded period.

### 5.4 Top 10 Regions with the Highest Average Unemployment Rate

```python
top_10 = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False).head(10).round(2)
top_10
```

```python
plt.figure(figsize=(11,6))
sns.barplot(x=top_10.index,
            y= top_10.values,
            palette= "Set1")
plt.title("Top 10 Regions by Average Unemployment Rate")
plt.xlabel("Regions")
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()
for i in range(len(top_10)):
    plt.text(i,top_10.values[i]+0.5,str(top_10.values[i]),ha = 'center')
plt.show()
```

**insight :**

- Tripura has the highest average unemployment rate (28.35%) among all regions in the dataset.
- Haryana (26.28%) and Jharkhand (20.58%) also recorded significantly high average unemployment rates.
- Bihar and Himachal Pradesh have moderately high unemployment rates, indicating employment challenges.
- Uttar Pradesh (12.55%) has the lowest average unemployment rate among the top 10 regions.
- The analysis shows that unemployment levels vary considerably across different regions, highlighting regional differences in employment opportunities.

### 5.5 Average Unemployment Rated by Area 

```python
avg_area = df.groupby("Area")["Estimated Unemployment Rate (%)"].mean()
plt.Figure(figsize=(10,5))
sns.barplot(x= avg_area.index,
            y= avg_area.values,
            palette="Set2")
plt.title("Average unemployment Rate by Area")
plt.xlabel("Area")
plt.grid(True)
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()
plt.show()
```

**insight:**
Urban areas have a higher unemployment rate than rural areas.

### 5.6 Monthly Average Unemployment Rate

```python
df["Month"] = df["Date"].dt.month_name()
```

```python
monthly_avg = df.groupby("Month")["Estimated Unemployment Rate (%)"].mean()
```

```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

monthly_avg = monthly_avg.reindex(month_order)

plt.figure(figsize=(10,5))
monthly_avg.plot(kind="line", marker="o")
plt.title("Monthly Average Unemployment Rate")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")
plt.grid(True)
plt.show()
```

**insight:**
Month between March and July tend higher Average unemployment rate(%)

### 5.7 Covid-19 impact on Unemployment

```python
sns.lineplot(data = df,
             x= "Date",
             y="Estimated Unemployment Rate (%)")
plt.grid(True)
plt.title("Covid-19 Impact")
```

**insight :**
- Unemployment remained relatively stable before 2020.
- A sharp increase was observed during the COVID-19 period.
- The unemployment rate decreased again after the peak.

### 5.8 Correlation Heatmap

```python
plt.Figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include="number").corr(),
            annot= True,
            cmap= "coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

**insight :**
- Unemployment Rate and Estimated Employed have a weak negative correlation.
- Labour Participation Rate has very weak correlation with the other variables.
- No strong linear relationship exists among these numerical features.

### 5.9 Distribution of Unemployment Rate

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Estimated Unemployment Rate (%)"],
             bins = 20,
             kde = True)
plt.title("Distribution on Unemployment Rate")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Frequency")
plt.show()
```

**insight :**
- Most unemployment rates are below 20%.
- The distribution is positively (right) skewed.
- A few extreme values (outliers) are present.
